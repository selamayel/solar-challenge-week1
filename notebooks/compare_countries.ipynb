```python
import pandas as pd

benin = pd.read_csv("../data/benin_clean.csv")
sl = pd.read_csv("../data/sierra_leone_clean.csv")
togo = pd.read_csv("../data/togo_clean.csv")

benin["country"] = "Benin"
sl["country"] = "Sierra Leone"
togo["country"] = "Togo"

df = pd.concat([benin, sl, togo], ignore_index=True)

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))
for i, col in enumerate(["GHI", "DNI", "DHI"]):
    plt.subplot(1, 3, i+1)
    sns.boxplot(data=df, x="country", y=col)
    plt.title(f"{col} by Country")
plt.tight_layout()
plt.show()

summary = df.groupby("country")[["GHI", "DNI", "DHI"]].agg(["mean", "median", "std"])
summary

```
